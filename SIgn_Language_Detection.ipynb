```python
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break
    else:
      cv2.imshow('frame', frame)
      if cv2.waitKey(25) & 0xFF == 13:
        break
    
cap.release()
cv2.destroyAllWindows()    
```
